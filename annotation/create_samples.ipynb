```python
import pandas as pd
import os
#TODO: Get the updated netflix_reviews
#TODO: also fetch prime
df_n = pd.read_csv("../content_libraries/netflix_reviews_tmp.csv")
```

```python
sample_size_per_stratum = 1000
RANDOM_STATE = 42

sampled_df_n = (
    df_n.groupby(["year"], group_keys=False)
      .sample(
          n=sample_size_per_stratum,
          random_state=RANDOM_STATE
      )
      .reset_index(drop=True)
)

sampled_df_n = sampled_df_n.drop(columns=["content", "thumbsUpCount", "at", "word_count"], errors="ignore")

print("Total sampled netflix reviews:", len(sampled_df_n))

print(sampled_df_n.groupby(["year"]).size())
```

```python
output_dir = "../content_libraries/netflix_samples"
os.makedirs(output_dir, exist_ok=True)

for year, group in sampled_df_n.groupby("year"):
    group.to_csv(
        os.path.join(output_dir, f"netflix_reviews_{year}_sample.csv"),
        index=False
    )
```
