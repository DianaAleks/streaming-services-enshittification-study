```python
# Genre information for the first few titles
netflix["genres"].head()
```

```python
import ast

netflix["genres"] = netflix["genres"].apply(ast.literal_eval)
```

```python
netflix["genre_names"] = netflix["genres"].apply(
    lambda genres: [genre["name"] for genre in genres]
)
```

```python
netflix_genres = netflix["genre_names"].explode()

netflix_genres.value_counts()
```

```python
# Genre information for the first few titles
amazon["genres"].head()
```

```python
amazon["genres"] = amazon["genres"].apply(ast.literal_eval)

amazon["genre_names"] = amazon["genres"].apply(
    lambda genres: [genre["name"] for genre in genres]
)

amazon_genres = amazon["genre_names"].explode()

amazon_genres.value_counts()
```

```python
# Top 10 Netflix genres
top_netflix_genres = netflix_genres.value_counts().head(10)
top_netflix_genres.plot(kind="bar")
plt.title("Top 10 Netflix Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
# Top 10 Amazon Prime genres
top_amazon_genres = amazon_genres.value_counts().head(10)
top_amazon_genres.plot(kind="bar")
plt.title("Top 10 Amazon Prime Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```
