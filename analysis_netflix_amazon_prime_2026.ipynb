```python
# Reading the IMDb ratings data
imdb = pd.read_csv("C:\\Users\\kamik\\Desktop\\Kamelija\\Документи\\Denmark\\Master Degree\\Classes\\Data in the Wild\\Project\\streaming-services-enshittification-study\\Title_ratings.tsv", sep="\t")

print(imdb.head())
print(imdb.columns.tolist())
```

```python
# Renaming columns for clarity
imdb = imdb.rename(columns={
    "tconst": "imdb_id",
    "averageRating": "imdb_score",
    "numVotes": "imdb_votes"
})

print(imdb.head())
```

```python
# Preparing the data for merging Netflix and IMDb datasets
netflix["imdb_id"] = netflix["imdb_id"].astype("string").str.strip()
imdb["imdb_id"] = imdb["imdb_id"].astype("string").str.strip()

netflix_rated = netflix.merge(
    imdb[["imdb_id", "imdb_score", "imdb_votes"]],
    on="imdb_id",
    how="left",
    validate="many_to_one"
)

print(netflix_rated[
    ["title", "imdb_id", "imdb_score", "imdb_votes"]
].head(10))
```

```python
# Preparing the data for merging Amazon Prime and IMDb datasets
amazon["imdb_id"] = amazon["imdb_id"].astype("string").str.strip()

amazon_rated = amazon.merge(
    imdb[["imdb_id", "imdb_score", "imdb_votes"]],
    on="imdb_id",
    how="left",
    validate="many_to_one"
)

print(amazon_rated[
    ["title", "imdb_id", "imdb_score", "imdb_votes"]
].head(10))
```

```python
# Checking for missing IMDb ratings in both datasets
for name, data in [
    ("Netflix", netflix_rated),
    ("Amazon Prime", amazon_rated)
]:
    print(f"\n{name}")
    print("Total rows:", len(data))
    print("Matched IMDb ratings:", data["imdb_score"].notna().sum())
    print("Missing IMDb ratings:", data["imdb_score"].isna().sum())
```

```python
# Filtering Drama titles with IMDb ratings
netflix_drama = netflix_rated[
    netflix_rated["genre_names"].apply(lambda genres: "Drama" in genres)
].copy()

amazon_drama = amazon_rated[
    amazon_rated["genre_names"].apply(lambda genres: "Drama" in genres)
].copy()
```

```python
netflix_drama["platform"] = "Netflix"
amazon_drama["platform"] = "Amazon Prime"
```

```python
# Combining Drama titles from both platforms
all_drama = pd.concat(
    [netflix_drama, amazon_drama],
    ignore_index=True
)

valid_drama = all_drama.dropna(
    subset=["imdb_score", "imdb_votes"]
).copy()
```

```python
# Comparing platforms based on Drama titles
drama_rating_stats = valid_drama.groupby("platform")["imdb_score"].agg(
    mean="mean",
    median="median",
    count="count"
).round(2)

print("IMDb ratings for Drama by platform:")
print(drama_rating_stats)
```

```python
# Summary statistics for Drama titles on each platform
for name, data in [
    ("Netflix", netflix_drama),
    ("Amazon Prime", amazon_drama)
]:
    print(f"\n{name}")
    print("Drama titles:", data["tmdb_id"].nunique())
    print("Missing IMDb ratings:", data["imdb_score"].isna().sum())
    print("Missing vote counts:", data["imdb_votes"].isna().sum())
    print(
        "IMDb rating coverage (%):",
        round(data["imdb_score"].notna().mean() * 100, 2)
    )
```

```python
# Visualizing the release-year distribution for Drama titles on each platform
for name, data in [
    ("Netflix", netflix_drama),
    ("Amazon Prime", amazon_drama)
]:
    data = data.copy()
    data["release_year"] = pd.to_numeric(
        data["release_year"], errors="coerce"
    )

    print(f"\n{name}")
    print(data["release_year"].describe().round(1))

    data["release_year"].dropna().hist(bins=20, figsize=(8, 4))
    plt.title(f"Release-Year Distribution — {name} (2026)")
    plt.xlabel("Release year")
    plt.ylabel("Number of titles")
    plt.tight_layout()
    plt.show()
```

```python
# Adding age group based on release year
def add_age_group(data, snapshot_year=2026):
    data = data.copy()
    data["release_year"] = pd.to_numeric(
        data["release_year"], errors="coerce"
    )

    data["age_group"] = "Unknown"
    data.loc[
        data["release_year"].between(snapshot_year - 4, snapshot_year),
        "age_group"
    ] = "Recent (last 5 years)"

    data.loc[
        data["release_year"] < snapshot_year - 4,
        "age_group"
    ] = "Older"

    return data

netflix_drama = add_age_group(netflix_drama)
amazon_drama = add_age_group(amazon_drama)

age_data = pd.concat([
    netflix_drama.assign(platform="Netflix"),
    amazon_drama.assign(platform="Amazon Prime")
], ignore_index=True)

age_counts = age_data.groupby(
    ["platform", "age_group"]
)["tmdb_id"].nunique().unstack(fill_value=0)

print(age_counts)

age_counts.plot(kind="bar", stacked=True, figsize=(8, 5), rot=0)
plt.title("Recent vs Older Drama Titles (2026)")
plt.xlabel("Platform")
plt.ylabel("Number of unique titles")
plt.tight_layout()
plt.show()
```

```python
# Only titles with a positive, recorded vote count
positive_votes = all_drama[
    all_drama["imdb_votes"] > 0
].copy()

plt.figure(figsize=(10, 5))

for platform in ["Netflix", "Amazon Prime"]:
    votes = positive_votes.loc[
        positive_votes["platform"] == platform,
        "imdb_votes"
    ]
    plt.hist(
        votes,
        bins=50,
        alpha=0.5,
        label=platform
    )

plt.xscale("log")
plt.xlabel("IMDb vote count (log scale)")
plt.ylabel("Number of catalogue entries")
plt.title("Distribution of IMDb Votes for Drama Titles (2026)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Age-adjusted comparison of Drama titles by platform
rating_data = valid_drama.copy()

rating_data["release_year"] = pd.to_numeric(
    rating_data["release_year"], errors="coerce"
)

rating_data["release_period"] = pd.cut(
    rating_data["release_year"],
    bins=[1900, 1999, 2009, 2019, 2026],
    labels=["Before 2000", "2000–2009", "2010–2019", "2020–2026"]
)

age_adjusted_comparison = rating_data.groupby(
    ["platform", "release_period"], observed=False
).agg(
    titles=("tmdb_id", "nunique"),
    mean_imdb_rating=("imdb_score", "mean"),
    median_imdb_rating=("imdb_score", "median")
).round(2)

print(age_adjusted_comparison)
```

```python
# Content quality based on available titles
content_quality = all_titles.groupby("platform").agg(
    total_catalogue_rows=("title", "size"),
    rated_titles=("imdb_score", "count"),
    mean_imdb_rating=("imdb_score", "mean"),
    median_imdb_rating=("imdb_score", "median")
).round(2)

content_quality["rating_coverage_pct"] = (
    content_quality["rated_titles"]
    / content_quality["total_catalogue_rows"]
    * 100
).round(2)

print(content_quality)
```

```python
# High-quality titles: IMDb rating >= 7 and more than 7,000 votes
all_titles = all_titles.copy()

all_titles["high_quality"] = (
    (all_titles["imdb_score"] >= 7) &
    (all_titles["imdb_votes"] > 7000)
)

# Compare Netflix and Amazon Prime across all genres
quality_comparison = all_titles.groupby("platform").agg(
    total_catalogue_titles=("title", "size"),
    titles_with_imdb_rating=("imdb_score", "count"),
    high_quality_titles=("high_quality", "sum"),
    average_imdb_rating=("imdb_score", "mean"),
    median_imdb_rating=("imdb_score", "median")
)

# Share of the entire catalogue that meets both criteria
quality_comparison["high_quality_share_all_titles_pct"] = (
    quality_comparison["high_quality_titles"] /
    quality_comparison["total_catalogue_titles"] * 100
)

# Share of titles with IMDb ratings that meet both criteria
quality_comparison["high_quality_share_rated_titles_pct"] = (
    quality_comparison["high_quality_titles"] /
    quality_comparison["titles_with_imdb_rating"] * 100
)

print(quality_comparison.round(2))
```

```python
# Split titles into individual genres
genre_data = all_titles.dropna(
    subset=["genre_names", "imdb_score", "platform"]
).copy()

genre_data["genre"] = genre_data["genre_names"].str.split(",")
genre_data = genre_data.explode("genre")
genre_data["genre"] = genre_data["genre"].str.strip()

# Average IMDb score and number of titles for each genre/platform
genre_ratings = (
    genre_data.groupby(["platform", "genre"])
    .agg(
        average_imdb_score=("imdb_score", "mean"),
        number_of_titles=("imdb_score", "count")
    )
    .reset_index()
)

# Sort from highest to lowest average IMDb score within each platform
genre_ratings = genre_ratings.sort_values(
    ["platform", "average_imdb_score"],
    ascending=[True, False]
)

# Show the top 5 genres for each platform
top_5_genres = genre_ratings.groupby("platform").head(5)

print(top_5_genres.round(2).to_string(index=False))
```

```python
# Standardize IDs
all_titles["imdb_id"] = (
    all_titles["imdb_id"].astype("string").str.strip()
)
imdb_titles["tconst"] = (
    imdb_titles["tconst"].astype("string").str.strip()
)

# Match all streaming titles to IMDb by ID
all_comparison = all_titles.merge(
    imdb_titles[["tconst", "primaryTitle"]],
    left_on="imdb_id",
    right_on="tconst",
    how="inner"
)

# Compare titles, ignoring capitalization and surrounding spaces
all_comparison["same_title"] = (
    all_comparison["title"].fillna("").str.strip().str.casefold()
    ==
    all_comparison["primaryTitle"].fillna("").str.strip().str.casefold()
)

# Overall results
print("All streaming titles:", len(all_titles))
print("Matched IMDb IDs:", len(all_comparison))
print("Same title:", all_comparison["same_title"].sum())
print("Different titles:", (~all_comparison["same_title"]).sum())

if len(all_comparison) > 0:
    print(
        "Exact title match percentage:",
        round(all_comparison["same_title"].mean() * 100, 2), "%"
    )
```

```python
platform_comparison = all_comparison.groupby("platform").agg(
    matched_titles=("imdb_id", "size"),
    exact_title_matches=("same_title", "sum")
)

platform_comparison["different_titles"] = (
    platform_comparison["matched_titles"]
    - platform_comparison["exact_title_matches"]
)

platform_comparison["exact_match_percentage"] = (
    platform_comparison["exact_title_matches"]
    / platform_comparison["matched_titles"] * 100
).round(2)

print(platform_comparison)
```
