```python
# Reading the IMDb ratings data
imdb = pd.read_csv("C:\\Users\\kamik\\Desktop\\Kamelija\\Документи\\Denmark\\Master Degree\\Classes\\Data in the Wild\\Project\\streaming-services-enshittification-study\\Title_ratings.tsv", sep="\t")

print(imdb.head())
print(imdb.columns.tolist())
```

```python
# Renaming columns for clarity
imdb = imdb.rename(columns={
    "tconst": "imdb_id",
    "averageRating": "imdb_score",
    "numVotes": "imdb_votes"
})

print(imdb.head())
```

```python
# Preparing the data for merging Netflix and IMDb datasets
netflix["imdb_id"] = netflix["imdb_id"].astype("string").str.strip()
imdb["imdb_id"] = imdb["imdb_id"].astype("string").str.strip()

netflix_rated = netflix.merge(
    imdb[["imdb_id", "imdb_rating", "num_votes"]],
    on="imdb_id",
    how="left",
    validate="many_to_one"
)

print(netflix_rated[
    ["title", "imdb_id", "imdb_rating", "num_votes"]
].head(10))
```

```python
# Preparing the data for merging Amazon Prime and IMDb datasets
amazon["imdb_id"] = amazon["imdb_id"].astype("string").str.strip()

amazon_rated = amazon.merge(
    imdb[["imdb_id", "imdb_rating", "num_votes"]],
    on="imdb_id",
    how="left",
    validate="many_to_one"
)

print(amazon_rated[
    ["title", "imdb_id", "imdb_rating", "num_votes"]
].head(10))
```

```python
# Checking for missing IMDb ratings in both datasets
for name, data in [
    ("Netflix", netflix_rated),
    ("Amazon Prime", amazon_rated)
]:
    print(f"\n{name}")
    print("Total rows:", len(data))
    print("Matched IMDb ratings:", data["imdb_rating"].notna().sum())
    print("Missing IMDb ratings:", data["imdb_rating"].isna().sum())
```

```python
# Filtering Drama titles with IMDb ratings
netflix_drama = netflix_rated[
    netflix_rated["genre_names"].apply(lambda genres: "Drama" in genres)
].copy()

amazon_drama = amazon_rated[
    amazon_rated["genre_names"].apply(lambda genres: "Drama" in genres)
].copy()
```

```python
# Checking for missing IMDb ratings in Drama titles
netflix_drama["platform"] = "Netflix"
amazon_drama["platform"] = "Amazon Prime"
```

```python
# Combining Drama titles from both platforms
all_drama = pd.concat(
    [netflix_drama, amazon_drama],
    ignore_index=True
)

valid_drama = all_drama.dropna(
    subset=["imdb_rating", "num_votes"]
).copy()
```

```python
# Calculating the mean IMDb rating and the 75th percentile of the number of votes
C = valid_drama["imdb_rating"].mean()
m = valid_drama["num_votes"].quantile(0.75)

valid_drama["bayesian_rating"] = (
    valid_drama["num_votes"] / (valid_drama["num_votes"] + m)
    * valid_drama["imdb_rating"]
    +
    m / (valid_drama["num_votes"] + m) * C
)

print("Overall Drama rating:", round(C, 2))
print("Vote threshold:", round(m, 0))

print(valid_drama[
    ["title", "platform", "imdb_rating", "num_votes", "bayesian_rating"]
].head(10))
```

```python
# Comparing platforms based on Drama titles
platform_comparison = (
    valid_drama.groupby("platform")
    .agg(
        number_of_titles=("tmdb_id", "nunique"),
        average_imdb_rating=("imdb_rating", "mean"),
        average_bayesian_rating=("bayesian_rating", "mean"),
        median_bayesian_rating=("bayesian_rating", "median")
    )
    .round(2)
)

print(platform_comparison)
```

```python
# Visualizing the average IMDb and Bayesian ratings for Drama titles by platform
platform_comparison[
    ["average_imdb_rating", "average_bayesian_rating"]
].plot(kind="bar", figsize=(8, 5), rot=0)

plt.title("Drama Ratings: IMDb vs Bayesian IMDb (2026)")
plt.xlabel("Streaming platform")
plt.ylabel("Average rating")
plt.tight_layout()
plt.show()
```

```python
# Summary statistics for Drama titles on each platform
for name, data in [
    ("Netflix", netflix_drama),
    ("Amazon Prime", amazon_drama)
]:
    print(f"\n{name}")
    print("Drama titles:", data["tmdb_id"].nunique())
    print("Missing IMDb ratings:", data["imdb_rating"].isna().sum())
    print("Missing vote counts:", data["num_votes"].isna().sum())
    print(
        "IMDb rating coverage (%):",
        round(data["imdb_rating"].notna().mean() * 100, 2)
    )
```

```python
# Visualizing the release-year distribution for Drama titles on each platform
for name, data in [
    ("Netflix", netflix_drama),
    ("Amazon Prime", amazon_drama)
]:
    data = data.copy()
    data["release_year"] = pd.to_numeric(
        data["release_year"], errors="coerce"
    )

    print(f"\n{name}")
    print(data["release_year"].describe().round(1))

    data["release_year"].dropna().hist(bins=20, figsize=(8, 4))
    plt.title(f"Release-Year Distribution — {name} (2026)")
    plt.xlabel("Release year")
    plt.ylabel("Number of titles")
    plt.tight_layout()
    plt.show()
```

```python
# Adding age group based on release year
def add_age_group(data, snapshot_year=2026):
    data = data.copy()
    data["release_year"] = pd.to_numeric(
        data["release_year"], errors="coerce"
    )

    data["age_group"] = "Unknown"
    data.loc[
        data["release_year"].between(snapshot_year - 4, snapshot_year),
        "age_group"
    ] = "Recent (last 5 years)"

    data.loc[
        data["release_year"] < snapshot_year - 4,
        "age_group"
    ] = "Older"

    return data

netflix_drama = add_age_group(netflix_drama)
amazon_drama = add_age_group(amazon_drama)

age_data = pd.concat([
    netflix_drama.assign(platform="Netflix"),
    amazon_drama.assign(platform="Amazon Prime")
], ignore_index=True)

age_counts = age_data.groupby(
    ["platform", "age_group"]
)["tmdb_id"].nunique().unstack(fill_value=0)

print(age_counts)

age_counts.plot(kind="bar", stacked=True, figsize=(8, 5), rot=0)
plt.title("Recent vs Older Drama Titles (2026)")
plt.xlabel("Platform")
plt.ylabel("Number of unique titles")
plt.tight_layout()
plt.show()
```

```python
# Age-adjusted comparison of Drama titles by platform
rating_data = valid_drama.copy()
rating_data["release_year"] = pd.to_numeric(
    rating_data["release_year"], errors="coerce"
)

rating_data["release_period"] = pd.cut(
    rating_data["release_year"],
    bins=[1900, 1999, 2009, 2019, 2026],
    labels=["Before 2000", "2000–2009", "2010–2019", "2020–2026"]
)

age_adjusted_comparison = rating_data.groupby(
    ["platform", "release_period"], observed=False
).agg(
    titles=("tmdb_id", "nunique"),
    mean_bayesian_rating=("bayesian_rating", "mean")
).round(2)

print(age_adjusted_comparison)
```

```python
# Vote comparison of Drama titles by platform
vote_summary = all_drama.groupby("platform").agg(
    total_titles=("tmdb_id", "nunique"),
    missing_votes=("num_votes", lambda x: x.isna().sum()),
    zero_votes=("num_votes", lambda x: (x == 0).sum()),
    positive_votes=("num_votes", lambda x: (x > 0).sum())
)

print(vote_summary)
```

```python
# Only titles with a positive, recorded vote count
positive_votes = all_drama[
    all_drama["num_votes"] > 0
].copy()

plt.figure(figsize=(10, 5))

for platform in ["Netflix", "Amazon Prime"]:
    votes = positive_votes.loc[
        positive_votes["platform"] == platform,
        "num_votes"
    ]
    plt.hist(
        votes,
        bins=50,
        alpha=0.5,
        label=platform
    )

plt.xscale("log")
plt.xlabel("IMDb vote count (log scale)")
plt.ylabel("Number of catalogue entries")
plt.title("Distribution of IMDb Votes for Drama Titles (2026)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Vote distribution by platform
vote_distribution = all_drama.groupby("platform")["num_votes"].describe(
    percentiles=[0.25, 0.50, 0.75, 0.90, 0.95]
)

print(vote_distribution.round(2))
```

```python
# Analyzing rating stability across platforms
rating_data = all_drama.dropna(
    subset=["imdb_rating", "num_votes"]
).copy()

rating_data = rating_data[rating_data["num_votes"] > 0]

vote_bins = [0, 10, 50, 100, 250, 500, 1000, float("inf")]
vote_labels = [
    "1–10", "11–50", "51–100", "101–250",
    "251–500", "501–1000", "1000+"
]

rating_data["vote_group"] = pd.cut(
    rating_data["num_votes"],
    bins=vote_bins,
    labels=vote_labels,
    include_lowest=True
)

stability = rating_data.groupby(
    ["platform", "vote_group"], observed=False
).agg(
    number_of_titles=("imdb_rating", "size"),
    mean_rating=("imdb_rating", "mean"),
    median_rating=("imdb_rating", "median"),
    rating_std=("imdb_rating", "std")
).round(2)

print(stability)
```
